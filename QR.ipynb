```python
from PIL import Image, ImageEnhance
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image
from PIL import ImageDraw
from PIL import ImageFont
import qrcode
import pandas as pd
```

```python
fld = r'L:\LU\Staff\Ricky\---System Checking\QR\P'
#data = '11SW-DFR324'

slplist = r'L:\LU\Staff\Ricky\---System Checking\QR\P\slplist.xlsx'

dfslp = pd.read_excel(slplist)


#======xl
import xlsxwriter
xlpath = Path(fld, 'qr.xlsx')
workbook = xlsxwriter.Workbook(xlpath)
worksheet = workbook.add_worksheet()

#===

for index, row in dfslp.iterrows():
    data = row['Slope']

    path = Path(fld, data +'.png')

    #========option1
    #img = qrcode.make(data)

    #========option2
    qr = qrcode.QRCode(version = 1, box_size = 10, border = 3)

    # Adding data to the instance 'qr'
    qr.add_data(data)

    qr.make(fit = True)
    img = qr.make_image(fill_color = 'blue',
                        back_color = 'white')

    #=========
    img.save(path)

    #===
    img=Image.open(path)
    draw=ImageDraw.Draw(img)
    font = ImageFont.truetype("ariblk.ttf", size = 15)
    draw.text((12,2),str(data),font=font, fill='#000000')
    #img

    #====resize
    #newsize = (300, 300)
    #img = img.resize(newsize)

    #=====
    path = Path(fld, data +'.png')
    img.save(path)

    #=====xl
    worksheet.insert_image('B'+str(index), path)
    worksheet.write('A'+str(index), data)

workbook.close()
```
