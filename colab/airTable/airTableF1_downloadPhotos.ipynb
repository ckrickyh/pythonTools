```python
from pyairtable import Api
import requests
import os
import pandas as pd

#from google.colab import drive

#google
#drive.mount('/content/drive')
os.makedirs(drive_path, exist_ok=True)
```

```python
api = Api(token)
table = api.table(baseId,tableName)
records = table.all()
```

# download photos to gdrive

```python
# directory = 'airTablePhoto'
# if not os.path.exists(directory):
#     os.makedirs(directory)

treeIds = []
newFileNames = []
counts = []
urls = []
successs = []

for record in records:
  #print(f'record: {record}')
  if 'treePhoto' in (record['fields']):
    #print(record['fields'])
    count = 0
    for i in record['fields']['treePhoto']:
      treeId = record['fields']['Tree No.']

      print(f'treeId: {treeId}')
      if count == 0:
        newFileName = record['fields']['Tree No.'] + '__Whole View'
      else:
        newFileName = record['fields']['Tree No.'] + '_' + 'd' + str(count)

      url = i['url']
      print(f'url: {url}')

      # ===== download files
      r = requests.get(url, newFileName) # r as response obj
      if r.status_code == 200:
        filePath = os.path.join(drive_path, f"{newFileName}.JPG")
        with open(filePath,'wb') as f:
           f.write(r.content)
           success = True
      else:
        success = False
        print(f'file cannot donwload {treeId}')


      # =============
      treeIds.append(treeId)
      newFileNames.append(newFileName)
      counts.append(count)
      urls.append(url)
      successs.append(success)

      count+=1

#===========
data = {
     'Tree No.': treeIds,
     'newFileName': newFileNames,
     'count': counts,
     'url': urls,
     'successs': success
 }
df = pd.DataFrame(data)
df_sorted = df.sort_values(by=['Tree No.', 'count'])
df_sorted.reset_index(drop=True, inplace=True)
df_sorted.to_csv(os.path.join(drive_path, "airTablePhoto.csv"))
print('done')
```
